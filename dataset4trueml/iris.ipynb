```python
import pandas as pd
from scipy.io import arff

data, meta = arff.loadarff('iris.arff')
df = pd.DataFrame(data)

# Clean up byte strings (SciPy loads nominal/text fields as bytes, e.g., b'sunny')
for col in df.columns:
    if df[col].dtype == object:
        df[col] = df[col].str.decode('utf-8')

df
```

```python
import numpy as np

feature_names = np.array(
    df.columns[:-1].tolist(),
    dtype="<U20"
)

X = df.drop(columns="class").to_numpy(dtype=np.float64)


target_names, y = np.unique(
    df["class"].to_numpy(dtype=str),
    return_inverse=True,

)

target_names = np.array(
    target_names.tolist(),
    dtype="<U20"
)

np.savez_compressed("iris.npz", 
                    data=X, target=y.astype(np.int64), 
                    feature_names=feature_names, 
                    target_names=target_names,
                   ) 
```

```python
with np.load("iris.npz", allow_pickle=True) as d:
    print(d.files)

    for name in d.files:
        print(name, d[name].shape, d[name].dtype)
```
