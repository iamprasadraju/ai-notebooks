```python
# Dataset: https://www.kaggle.com/datasets/uciml/iris
import numpy as np
import pandas as pd

df = pd.read_csv("csv-datasets/iris.csv")
df.head()
```

```python
X = df.drop(columns=["Species", "Id"])
y = df[["Species"]]


# native fixed-width unicode string array.
# fast, secure, compact data format. zero boilerplate loading (allow_pickle=True at npz loading)
input_features = np.array(X.columns, dtype="U")
target_features = np.array(y.columns, dtype="U")
target_classes = np.array(list(y["Species"].unique()), dtype="U")

print(input_features, target_features)
print(target_classes)

print("Target classes data type:", target_classes.dtype)
```

```python
data = X.to_numpy()

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
target = encoder.fit_transform(y.values.ravel())

mapping = {name: idx for idx, name in enumerate(encoder.classes_)}
print(mapping)
```

```python
np.savez_compressed(
    "iris.npz",
    data = data,
    target = target,
    input_features = input_features,
    target_features = target_features,
    target_classes = target_classes
)
```
