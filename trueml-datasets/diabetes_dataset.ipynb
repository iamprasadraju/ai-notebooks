```python
# Dataset: https://www4.stat.ncsu.edu/~boos/var.select/diabetes.tab.txt

import numpy as np
import pandas as pd


df = pd.read_csv("csv-datasets/diabetes.tab.txt", sep='\t')
print(df.head())
df.info()
```

```python
X = df.drop(columns=["Y"])
y = df[["Y"]]

# print(X.columns)
# print(y.columns)

input_features = np.array(X.columns, dtype="U")
target_features = np.array(y.columns, dtype="U")

print(input_features, target_features)
```

```python
data = X.to_numpy()
target = y.to_numpy().flatten()

print(data.shape, target.shape)
```

```python
np.savez_compressed(
    "diabetes.npz",
    data = data,
    target = target,
    input_features = input_features,
    target_features = target_features,
)
```
