```python
# Datasets: https://www.kaggle.com/datasets/uciml/breast-cancer-wisconsin-data

import numpy as np
import pandas as pd


df = pd.read_csv("csv-datasets/breast_cancer.csv")
df.info()
```

```python
X = df.drop(columns=["diagnosis", "id", "Unnamed: 32"])
y = df[["diagnosis"]]

#print(X.columns)
#print(y.columns)

input_features = np.array(X.columns, dtype="U")
target_features = np.array(y.columns, dtype="U")
traget_classes = np.array(y["diagnosis"].unique(), dtype="U")


print(input_features," - ", target_features)
print(target_classes)
```

```python
data = X.to_numpy()

from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
target = encoder.fit_transform(y.values.ravel())

mapping = {name: idx for idx, name in enumerate(encoder.classes_)}
print(mapping)
```

```python
np.savez_compressed(
    "breast_cancer.npz",
    data = data,
    target = target,
    input_features = input_features,
    target_features = target_features,
    target_classes = target_classes
)
```
