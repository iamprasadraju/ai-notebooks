```python
# trueml need this dataset (binary)
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler

data = load_breast_cancer()
X, y = data.data, data.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
print(y.shape)
```

```python
from trueml.linear_model import LinearModel
from trueml.activations import Sigmoid
import numpy as np


model = LinearModel(n_features=X_scaled.shape[1], n_targets=1)
activ_fn = Sigmoid()

y_pred = []
for epoch in range(1000):
    res = model.forward(X_scaled)
    y_pred = activ_fn(res)
    break
    

for i in range(100):
    with np.printoptions(suppress=True):
        print("Y_pred: ", y_pred[i], "  Y_true: ", y[i]) 
```

### Designing a loss function (Binary Cross Entropy)
---

Assume: $$Loss = \Sigma(Ytrue - Ypred)$$

Example: \(1 - 0.00001578 = 0.99998422\). What is wrong with this? (Why this is not a good loss function in this case)

Reasons: 

- This generates positive and negative errors, which leads to cancellation when adding up the total loss.
- It does not heavily penalize large errors (the maximum error is capped at 1, which is not enough motivation for the model to learn). (apply same for MSE)


### How do we amplify large errors? Does such a function exist?

Yes, we can use a negative logarithm function, but we do not put the subtraction directly inside the log. Instead, we take the negative log of the predicted probability for the correct class:

- When the true label is 1: $-\log(Y_{\text{pred}})$

- When the true label is 0: $-\log(1 - Y_{\text{pred}})$
