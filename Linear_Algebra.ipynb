# Introduction to Linear Algebra

**Algebra : “How to solve this equation?”**

**Linear Algebra: "What is happening to the whole space?"**

For example, ordinary algebra:

$$2x + 3 = 7$$

You find:
    $x = 2$
Done.

Linear algebra takes something like:

$$Ax=b$$

and asks deeper questions:
- What does the matrix \(A\) do to a vector?
- Do solutions exist?
- Is the solution unique?
- What directions are preserved?
- What is the space of all solutions?

The key idea is:
- Vectors are the objects.
- Matrices are the transformations.
- Linear algebra studies the relationships between them.
---

## What is a Vector ?

A vector is a quantity with both size and direction.


```python
%matplotlib inline

import matplotlib.pyplot as plt

class VectorPlot:
    def __init__(self):
        self.fig, self.ax = plt.subplots(figsize=(4, 4))

        self.ax.set_xlim(-8, 8)
        self.ax.set_ylim(-8, 8)
        #self.ax.grid()
        self.ax.set_aspect("equal")


    def add(self, origin, vector):
        self.ax.quiver(
            origin[0], origin[1],
            vector[0], vector[1],
            angles="xy",
            scale_units="xy",
            scale=1,
            width=0.004
        )

    def show(self):
        display(self.fig)
```

```python
import numpy as np

origin = np.array([0, 0])

v = np.array([[1, 1], [-2, 2], [4, -7]])
plot = VectorPlot()

plot.add(origin, v[0])
```

```python
plot.add(origin, 5 * v[0])
plot.show()
```
