```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

class VectorPlot:
    def __init__(self):
        # Apply the ggplot style locally inside initialization
        plt.style.use('ggplot')
        
        self.fig, self.ax = plt.subplots(figsize=(5, 5))
        self.ax.set_aspect("equal")

    def add(self, origin, vector):
        endpoint = (origin[0] + vector[0], origin[1] + vector[1])

        mx = max(abs(origin[0]), abs(origin[1]),
                 abs(endpoint[0]), abs(endpoint[1]), 1) + 2
        self.ax.set_xlim(-mx, mx)
        self.ax.set_ylim(-mx, mx)
        self.ax.quiver(
            origin[0], origin[1],
            vector[0], vector[1],
            scale_units="xy",
            scale=1,
            angles="xy",
            width=0.005
        )
        # Endpoint marker
        self.ax.scatter(*endpoint, s=50, zorder=3)
    
        # Vector label
        self.ax.annotate(
            f"[{vector[0]}, {vector[1]}]",
            endpoint,
            xytext=(5, 5),
            textcoords="offset points"
        )

    def show(self): 
        display(self.fig)

    def clear(self):
        self.ax.clear()
        self.ax.set_aspect("equal")
```

# Introduction to Linear Algebra

**Algebra : “How to solve this equation?”**

**Linear Algebra: "What is happening to the whole space?"**

For example, ordinary algebra:

$$2x + 3 = 7$$

You find:
    $x = 2$
Done.

Linear algebra takes something like:

$$Ax=b$$

and asks deeper questions:
- What does the matrix A do to a vector?
- Do solutions exist?
- Is the solution unique?
- What directions are preserved?
- What is the space of all solutions?

The key idea is:
- Vectors are the objects.
- Matrices are the transformations.
- Linear algebra studies the relationships between them.
---

**n linear equations, n unknowns**


## What is a Vector ?

A vector is a quantity with both size and direction. i.e \begin{bmatrix} x_1 \\ y_1 \end{bmatrix}

```python
## Vector on co-ordinate system
v = VectorPlot()
v1 = np.array([-2, 3])
origin = np.array([0, 0])

v.add(origin,  v1)
```

## Vector addition

Adding vectors combines their magnitudes and directions to find a single net value called the resultant vector.

Example: $\begin{bmatrix} 1 \\ 2 \end{bmatrix} + \begin{bmatrix} -2 \\ 3 \end{bmatrix} = \begin{bmatrix} -1 \\ 5 \end{bmatrix}$



```python
### vector addition
v.clear()
v2 = np.array([1, 2])
v3 = np.array([-2, 3])

v.add(origin, v2)
v.add(origin, v3)
v.add(origin, v2 + v3)
v.show()
```

## Scalar Multiplication (scaling)

the process of multiplying a vector or a matrix by a single number (called a scalar), which scales every component or entry by that value.

Example: $3 * \begin{bmatrix} -1 \\ 4 \end{bmatrix} = \begin{bmatrix} -3 \\ 12 \end{bmatrix}$

```python
s = 3
v4 = np.array([-1, 2])

v.clear()
v.add(origin, v4)
v.add(origin, 3 * v4)
v.add(origin, -2 * v4)
v.show()
```
